```python
# PROGRAMME D'OPTIMISATION DE CODONS (HOMO SAPIENS)


def optimiser_sequence(sequence_brute):

    # - ÉTAPE 1 : NOS DICTIONNAIRES DE TRADUCTION -
    meilleurs_codons_adn = {
        'A': 'GCC', 'C': 'TGC', 'D': 'GAC', 'E': 'GAG', 'F': 'TTC',
        'G': 'GGC', 'H': 'CAC', 'I': 'ATC', 'K': 'AAG', 'L': 'CTG',
        'M': 'ATG', 'N': 'AAC', 'P': 'CCC', 'Q': 'CAG', 'R': 'AGA',
        'S': 'AGC', 'T': 'ACC', 'V': 'GTG', 'W': 'TGG', 'Y': 'TAC', '*': 'TGA'
    }

    traduction_3_vers_1 = {
        'ALA': 'A', 'CYS': 'C', 'ASP': 'D', 'GLU': 'E', 'PHE': 'F',
        'GLY': 'G', 'HIS': 'H', 'ILE': 'I', 'LYS': 'K', 'LEU': 'L',
        'MET': 'M', 'ASN': 'N', 'PRO': 'P', 'GLN': 'Q', 'ARG': 'R',
        'SER': 'S', 'THR': 'T', 'VAL': 'V', 'TRP': 'W', 'TYR': 'Y'
    }

    # --- ÉTAPE 2 : NETTOYAGE DE LA SÉQUENCE ---
    # On met tout en majuscules pour éviter les erreurs de saisie
    sequence = sequence_brute.upper()
    lettres_autorisees = "ABCDEFGHIJKLMNOPQRSTUVWXYZ "
    sequence_propre = ""

    # On remplace tout ce qui n'est pas une lettre (virgules, tirets) par un espace
    for caractere in sequence:
        if caractere in lettres_autorisees:
            sequence_propre = sequence_propre + caractere
        else:
            sequence_propre = sequence_propre + " "

    # On découpe le texte à chaque espace (ex: "ALA GLY" devient ["ALA", "GLY"])
    mots = sequence_propre.split()

    if len(mots) == 0:
        return "Erreur : Vous n'avez rien écrit d'utilisable."

    # --- ÉTAPE 3 : CONVERSION TOUT EN 1 LETTRE ---
    acides_amines_1_lettre = []

    # Si le premier bloc fait 3 lettres, l'utilisateur a utilisé le format 3 lettres
    if len(mots[0]) == 3:
        for mot in mots:
            if mot in traduction_3_vers_1:
                acides_amines_1_lettre.append(traduction_3_vers_1[mot])
            else:
                return "Erreur : '" + mot + "' n'est pas reconnu."

    # Sinon, on considere que c'est le format 1 lettre
    else:
        # On recolle tout ensemble pour effacer les espaces éventuels
        chaine_collee = "".join(mots)
        for lettre in chaine_collee:
            acides_amines_1_lettre.append(lettre)

    # --- ÉTAPE 4 : TRADUCTION EN ADN ---
    sequence_adn = []

    for acide_amine in acides_amines_1_lettre:
        if acide_amine in meilleurs_codons_adn:
            codon = meilleurs_codons_adn[acide_amine]
            sequence_adn.append(codon)
        else:
            return "Erreur : La lettre '" + acide_amine + "' n'existe pas en biologie."

    # --- ÉTAPE 5 : PRÉSENTATION DU RÉSULTAT ---
    adn_avec_espaces = " ".join(sequence_adn)
    adn_tout_colle = "".join(sequence_adn)

    return "Format codons : " + adn_avec_espaces + "\nFormat collé  : " + adn_tout_colle



print("  Entrez la séquence protéique à traduire ")

# Le programme met l'exécution en pause et attend ta saisie dans la console
sequence_utilisateur = input("Collez votre séquence d'acides aminés ici, puis appuyez sur Entrée : ")

# Le programme calcule la séquence
resultat_final = optimiser_sequence(sequence_utilisateur)

# Le programme affiche le résultat final
print("\n RÉSULTAT :")
print(resultat_final)
```
